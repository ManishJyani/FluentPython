First Concept : Public vs Private  
In first chapter of fluent python , In FrenchDeck class I saw object attribute "self._cards", that made me curious about what is this underscore even mean?
From there I got to know the concept of public vs private attributes and methods.

```python
"""
Level 1 
self._cards = []
Meaning: "This is internal / private by convention"
Actually private? No — fully accessible from outside
Who respects it? Any developer who follows Python conventions
Extra behavior? One thing — if you do from module import *, names starting with _ are not imported
"""
```

```python
"""
Level 2 — Double Underscore __name (Name Mangling)
self.__cards = []
Meaning: "I really don't want you touching this"
Actually private? Still no — but Python mangles the name
What is name mangling? Python secretly renames __cards to _FrenchDeck__cards
So
deck.__cards      # AttributeError — name doesn't exist!
deck._FrenchDeck__cards  # Works — but now you're really forcing it
"""
```

```python
"""
Level 3 — Dunder __name__ (Double on Both Sides)
def __len__(self):
def __init__(self):
Meaning: These are Python's special methods — part of the data model
Not private at all — these are meant to be used by Python itself
Never name your own attributes like this — these are reserved for Python
"""
```

```python
"""
Why Does Python Do It This Way?
This is a deliberate philosophical choice. Python's philosophy is:

"We are all consenting adults here."

Python trusts developers. It doesn't lock things down. Instead it uses conventions and says:

If you see _name — respect it as private
If you really need to access it — you can, but you're on your own
If something breaks — that's on you

This is very different from Java's approach where the compiler enforces access control.
"""
```

```python
# Real word example
class BankAccount:
    
    def __init__(self, owner, balance):
        self.owner = owner           # Public attribute
        self._balance = balance      # Private attribute
        self.__transaction_key = 'X7K9'  # Strongly private

    # Public method — the interface you expose
    def deposit(self, amount):
        if self._is_valid_amount(amount):  # calls private method
            self._balance += amount
            self._log_transaction('deposit', amount)  # calls private method

    # Public method
    def withdraw(self, amount):
        if self._is_valid_amount(amount) and self._has_sufficient_funds(amount):
            self._balance -= amount
            self._log_transaction('withdrawal', amount)

    # Public method
    def get_balance(self):
        return self._balance

    # Private method — internal validation helper
    def _is_valid_amount(self, amount):
        return amount > 0

    # Private method — internal check
    def _has_sufficient_funds(self, amount):
        return self._balance >= amount

    # Private method — internal logging
    def _log_transaction(self, type, amount):
        print(f"Transaction: {type} of {amount}")
```

```python
# Local variable: defined inside a function and can be accessed only within that function.
# Enclosing variable: defined in an outer function and accessible inside its nested (inner) functions.
# Global variable: defined outside all functions and accessible throughout the module (including inside functions, unless shadowed).
```

```python
# local
def f():
    name= 'ManishJyani'
    print(name)
f()
print(name) #NameError if accessed outside of function
```

```python
# enclosing
def outer():
    a=10
    def inner():
        a=5 # only scope inside inner , can not alter value for outer function
        print(a)
    inner()
    print(a)

outer() 
```

```python
# encloing
def outer():
    a=10
    def inner():
        nonlocal a
        a=5 # Now a is nonlocal, says not to create a new 'a' varialbe, use 'a' of enclosing function and update it.
        print(a)
    inner()
    print(a)

outer() 
#if I don't define 'a' in outer, nonlocal line will give error
# enclosing function :function that contains another function inside it.
```

😒 Why getting this "UnboundLocalError" error , it should have accessed earlier defined name (which was defined outside program, and printed it)
Because

Compile time → Python sees s = inside the function, marks s as local in the function's scope. Global s is now shadowed/ignored for this function.

Runtime → print(s) executes, interpreter looks for s in local scope only, finds it exists but has no value yet → throws UnboundLocalError.



```python
"""
UnboundLocalError is a subclass of NameError, but the key detail is:

Python decides a variable is local if it sees an assignment to it anywhere in the function.

So even if the assignment happens later, Python treats it as local everywhere in that function.
"""
```

```python
"""
In UnboundLocalError — name exists in the symbol table, but not bound to an object:
s ──────► ???   (label exists, points to nothing)
In NameError — name doesn't even exist in the symbol table:
s              (label doesn't even exist)
"""
```

```python
""" 
More fundamental why Unbound, what does even mean?
The answer is in Python 
s = "hello"
We are binding the name s to the string object "hello" in memory. From this point, s is bound.
When you haven't done that assignment yet, s exists as a name in the local scope (compiler registered it), 
but it points to nothing. It is unbound

BOUND:              UNBOUND:
                    
s ──────► "hello"   s ──────► ???
(label pointing     (label exists,
 to something)       points to nothing)

This "binding" concept is actually core to Python's identity. In Python, variables are not containers like in C/Java —
they are just names bound to objects.
That's why Python uses the term name binding everywhere:

Assignment = → binding
import → binds a module to a name
def → binds a function to a name
for x in ... → binds each item to x

All of these are just different ways of binding a name to an object.
"""
```

```python
""" 
Now question comes how this binding is managed in python?
Answer : Symbol Table

A symbol table is a data structure maintained by the Python compiler that stores information about every name 
(variable, function, class etc.) it encounters in your code.
Think of it like a registry — when compiler sees a name, it registers it here along with metadata about it.

What does it store?
For each name it stores things like:

The name itself
Its scope (local, global, free)
Whether it's been assigned/bound
Whether it's been referenced
"""
```

```python
# just see this in action
import symtable

code = """
s = "hello"
def my_func():
    print(s)
    s = "world"
"""

table = symtable.symtable(code, "test", "exec")

# top level symbols
for sym in table.get_symbols():
    print(sym.get_name(), "| assigned:", sym.is_assigned())

# function level symbols
for child in table.get_children():
    print(f"\nInside {child.get_name()}:")
    for sym in child.get_symbols():
        print(sym.get_name(), "| local:", sym.is_local(), "| assigned:", sym.is_assigned())

code = """
s = "hello"
def my_func():
    print(s)
    s = "world"
"""

table = symtable.symtable(code, "test", "exec")

# top level symbols
for sym in table.get_symbols():
    print(sym.get_name(), "| assigned:", sym.is_assigned())

# function level symbols
for child in table.get_children():
    print(f"\nInside {child.get_name()}:")
    for sym in child.get_symbols():
        print(sym.get_name(), "| local:", sym.is_local(), "| assigned:", sym.is_assigned())
```

```python
# Now how to fix the problem 
# just  mark it as global variable
def f():
    global name
    print(name)
    name='Manish'
    print(name)

name="India"
print(name)
f()
print(name)
# Now python is no longer labelling name is local 
```

```python
"""
Source Code
    │
    ▼
Compiler scans → builds Symbol Table
    │                  │
    │            (name, scope,
    │             assigned?,
    │             referenced?)
    ▼
Bytecode generated (LOAD_FAST vs LOAD_GLOBAL based on symbol table)
    │
    ▼
Runtime executes bytecode
    │
    ▼
Looks up name in correct scope
    │
    ├── found & bound ✅ → works fine
    ├── found, not bound ❌ → UnboundLocalError
    └── not found anywhere ❌ → NameError
"""
```
